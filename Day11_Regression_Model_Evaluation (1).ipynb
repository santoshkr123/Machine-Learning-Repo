```python
Q1. What is MAE?

MAE is the average of the absolute differences between actual and predicted values.

Q2. Difference between MAE and MSE?

MAE uses absolute errors, whereas MSE squares errors. MSE penalizes large errors more strongly.

Q3. Why do we use RMSE?

RMSE converts MSE back to the original target unit and gives greater weight to larger errors.

Q4. Can R² be negative?

Yes. A negative R² indicates that the model performs worse than the baseline that predicts the mean of the actual target values.

Q5. Difference between R² and Adjusted R²?

R² generally does not decrease when predictors are added. Adjusted R² accounts for the number of predictors and penalizes unnecessary features.
```

```python
MAE : MAE is the average of the absolute differences between actual and predicted values.

MSE : MSE stands for Mean Squared Error.
It calculates the average of the squared differences between the actual values and the predicted values.

RMSE : RMSE stands for Root Mean Squared Error
It is the square root of MSE (Mean Squared Error).

R2_score  : Coefficiant Of determination
It tells us how well a regression model can explain the variation in the target variable compared to a baseline model that always predicts the mean of the target variable.

ADJUSTED RSQUARE : R² generally does not decrease when additional predictors are added to the model, even if they are not useful. Adjusted R² penalizes unnecessary features.
```

```python
Kaunsa metric kab use karein?
Metric	Kab useful hai?
MAE	Average error ko easily samajhna ho
MSE	Large errors ko strongly penalize karna ho
RMSE	Error ko original unit mein samajhna ho, large errors ko zyada weight dena ho
R²	Mean baseline ke comparison mein explained variation samajhni ho
Adjusted R²	Multiple predictors wale regression models compare karne hon
```
